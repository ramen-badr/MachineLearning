```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
%matplotlib inline
plt.rcParams['figure.figsize'] = (10, 6)

from sklearn.datasets import load_wine
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import OneHotEncoder
from sklearn.feature_selection import mutual_info_classif
from sklearn.discriminant_analysis import LinearDiscriminantAnalysis
from sklearn.svm import SVC
from category_encoders import MEstimateEncoder
```

## Задание 1

### Загрузка DataFrame

```python
wine_data = load_wine()

X = pd.DataFrame(wine_data.data, columns=wine_data.feature_names)
Y = wine_data.target
```

### Подготовка данных

```python
# Разбиваем на тренировочные и тестовые данные
X_train, X_test, Y_train, Y_test = train_test_split(X, Y, test_size=.3, random_state=0)
```

### Решение методом линейного дискриминанта

```python
lda = LinearDiscriminantAnalysis(solver='svd')
lda.fit(X_train, Y_train)

print('Точность на тренировочных данных {:.2f} %'.format(100*lda.score(X_train, Y_train)))
print('Точность на тестовых данных {:.2f} %'.format(100*lda.score(X_test, Y_test)))
```

### Решение методом SVM используя линейное ядро

```python
svml = SVC(kernel='linear')
svml.fit(X_train, Y_train)

print('Точность на тренировочных данных {:.2f} %'.format(100*svml.score(X_train, Y_train)))
print('Точность на тестовых данных {:.2f} %'.format(100*svml.score(X_test, Y_test)))
```

### Решение методом SVM используя квадратичное ядро

```python
svml = SVC(kernel='linear')
svml.fit(X_train, Y_train)

print('Точность на тренировочных данных {:.2f} %'.format(100*svml.score(X_train, Y_train)))
print('Точность на тестовых данных {:.2f} %'.format(100*svml.score(X_test, Y_test)))
```

### Подготовка данных

```python
# Вычисляем взаимную информацию
mi_scores = mutual_info_classif(X, Y, random_state=0)

# Порог для отбора признаков
threshold = 0.1

# Отбираем признаки
selected_features = X.columns[mi_scores > threshold]
X_reduced = X[selected_features]

# Разбиваем на тренировочные и тестовые данные
X_train, X_test, Y_train, Y_test = train_test_split(X_reduced, Y, test_size=.3, random_state=0)
```
